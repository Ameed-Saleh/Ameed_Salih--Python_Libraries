```python
import pandas as pd
import numpy  as np
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Task 1
# Draw the lines for:
# 1 -> x + y = 12 =>  (y = -x + 12)
# 2 -> x - y = 4  =>  (y = x -4)
# Add grid, legend, and axis lines

plt.figure(figsize=(12, 8))

def linear_function(x, m, b):
    y = m * x + b
    return y
x1_values = np.linspace(-15, 35, 20)
y1_values = -1 * x1_values + 12 # supported in np   
y2_values = x1_values - 4 # supported in np   

plt.axhline(0, color='black',linewidth=1.0, linestyle='--') # X-axis
plt.axvline(0, color='black',linewidth=1.0, linestyle='--') # Y-axis

plt.plot(x1_values, y1_values, color='green', label='y = -x + 12')
plt.plot(x1_values, y2_values, color='red', label='y = x - 4')
plt.legend()


# Task 2
# Use symbols, Eq, and solve to find the intersection point.
# Scatter the point and annotate it on the graph
from sympy import symbols, Eq, solve
x = symbols('x')
solution = solve( Eq(-x + 12, x - 4), symbols('x') )
x_junc = float(solution[0]) 
y_junc = x_junc - 4
plt.scatter(x_junc, y_junc, color='blue', zorder=5, s=150)
plt.annotate(f"({x_junc}, {y_junc})", (x_junc, y_junc), xytext=(x_junc -1.3, y_junc + 2), fontsize=14, color='blue')


# Task 3
# The label points are Y = np.array([2, 4, 8, 20])
# In the graph, the predicted Y values were Y_pred = np.array([2, 5, 10, 25])
# Calculate the MSE
Y = np.array([2, 4, 8, 20])
Y_pred = np.array([2, 5, 10, 25])

# 1->
errors = Y - Y_pred
squared_errors = errors ** 2
print(f'\nsquared_errors => {squared_errors}' )
mse = np.mean(squared_errors)
print("\nThe MSE is:", mse)
# 2->
mse = ((Y - Y_pred) ** 2).mean()
print("\nThe MSE is:", mse)


plt.xlim(-10, 20)
plt.ylim(-20, 25)
plt.title('Task 1/2/3', color='blue')
plt.grid(color='gray', linestyle='--',linewidth=0.5)


plt.show()
```
