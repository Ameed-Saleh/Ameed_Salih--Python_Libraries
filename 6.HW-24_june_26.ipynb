```python
# Task 1
# Setup & data: Load the file with df = pd.read_csv('Churn_Modelling.csv').
# Display the first 7 rows, then print df.shape and df.columns
df = pd.read_csv('Churn_Modelling.csv')
display(df.head(7))
print(f'\n df.shape -> {df.shape}\n')
print(f'df.columns -> {df.columns}')
```

```python
# Task 2
# Sorting: Sort customers by CreditScore from lowest to highest and show the first 7 rows
df = pd.read_csv('Churn_Modelling.csv')
display(df.sort_values(by=['CreditScore'], ascending=True).head(7))
```

```python
# Task 3
# Multi-column sorting: Sort by Geography ascending and Balance descending.
# Show the first 12 rows and explain what the two sort directions mean
df = pd.read_csv('Churn_Modelling.csv')
display(df.sort_values(by=['Geography', 'Balance'], ascending=[True, False]).head(12))
# Geography ascending -> lowest to highest this means sorting alphabetically (A to Z).
# Balance descending -> highest to lowest this means sorting numerically decreasing.
```

```python
# Task 4
# Counts: Use value_counts() to count customers by Geography.
# Then count each Geography + Gender combination
df = pd.read_csv('Churn_Modelling.csv')
display(df['Geography'].value_counts())
print()
display(df[['Geography', 'Gender']].value_counts())
```

```python
# Task 5
# Unique values: Show the unique values in Geography, Gender, NumOfProducts, and Exited.
# Then use nunique() to count how many different values each of these columns has
df = pd.read_csv('Churn_Modelling.csv')
columns = ['Geography', 'Gender', 'NumOfProducts', 'Exited']

for col in columns:
    display(f'{col} unique values: {df[col].unique()}')
print()
for col in columns:
    display(f'{col} nunique values: {df[col].nunique()}')    
```

```python
# Task 6
# Replace & map: Create a copy of the DataFrame.
# In the copy, replace Female with F and Male with M using .replace().
# Then solve the same change with .map()
df_copy_replace = df.copy()
df_copy_replace['Gender'] = df_copy_replace['Gender'].replace(['Female', 'Male'], ['F', 'M'])
display(df_copy_replace['Gender'].to_frame())
print()
df_copy_map = df.copy()
df_copy_map['Gender'] = df_copy_map['Gender'].map({'Female': 'F', 'Male': 'M'})
display(df_copy_map['Gender'].to_frame())
```

```python
# Task 7
# Top & lowest values: Find the 6 customers with the highest Balance using nlargest.
# Then find the 4 customers with the lowest CreditScore using nsmallest
df = pd.read_csv('Churn_Modelling.csv') # shows the sex highest Balace.
display(df.nlargest(6, 'Balance'))
print()
display(df.nsmallest(4, 'CreditScore')) # shows the four lowest CreditScore.
```

```python
# Task 8
# sample(): Show 9 random customers using sample(9).
# Then show a 3% random sample using sample(frac=0.03)
df = pd.read_csv('Churn_Modelling.csv')
display(df.sample(9))
print()
display(df.sample(frac=0.03))
```

```python
# Task 9
# Duplicates: Check whether the dataset has duplicated CustomerId values.
# Then use sorting plus drop_duplicates('Geography') to keep one customer per country with the highest Balance

df = pd.read_csv('Churn_Modelling.csv')
print(f'Are there duplicated CustomerIds? {df['CustomerId'].duplicated().any()}')
display(df['CustomerId'].duplicated().to_frame())
display(df.sort_values(by= ['Balance'], ascending=[False]).drop_duplicates(subset=['Geography'], keep='first'))
```

```python
# Task 10
# Largest values plus unique values: Find the 12 customers with the highest EstimatedSalary.
# From those rows, print the unique Geography values

df = pd.read_csv('Churn_Modelling.csv')
display(df.nlargest(12, 'EstimatedSalary'))
print()
display(df.nlargest(12, 'EstimatedSalary')['Geography'].unique())
```

```python
# Task 11
# Missing data: Use the DataFrame below and implement the requested cleaning operations

df_missing = pd.DataFrame({
    'A': [None, 2, 3, None, 5, None],
    'B': [10, None, None, 40, 50, None],
    'C': [100, None, 300, 400, None, None]
})
display(df_missing)
print('Task 11 -> Q1')
# Eliminate any row in the DataFrame that contains at least one NaN value
display(df_missing.dropna())
print('\nTask 11 -> Q2')
# Modify the previous task and eliminate only rows that have NaN values across all columns
display(df_missing.dropna(thresh=1))
print('\nTask 11 -> Q3')
# Eliminate rows that have at least one NaN in columns A and C only
display(df_missing.dropna(subset=['A', 'C']))
print('\nTask 11 -> Q4')
# Replace all NaN values with a specified value of your choice
df_filled = df_missing.fillna(1)
display(df_filled)
print('\nTask 11 -> Q5')
# Replace all NaN values in column B with the mean of the non-NaN values in that column
df_missing['B'] = df_missing['B'].fillna(df_missing['B'].mean())
display(df_missing)
```
