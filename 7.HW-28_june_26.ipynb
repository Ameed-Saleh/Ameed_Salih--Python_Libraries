```python
import pandas as pd
import numpy  as np
from IPython.display import display
```

```python
df = pd.read_csv('tips2.csv')
display(df)
```

```python
# Task 1
# Load tips.csv. Display the first rows, df.info(), and df.describe().T. 
#Then add a new column price_per_person calculated as total_bill / size.

display(df.info())
print()
display(df.describe().T)
print()
df['price_per_person '] = df['total_bill'] / df['size']
display(df)
```

```python
# Task 2
# Count how many orders (rows) were placed at each time and for each gender.
df[['time', 'sex']].value_counts().sort_index().to_frame()
```

```python
# Task 3
# Count how many orders (rows) were placed on each day.
df['day'].value_counts().sort_index().to_frame()
```

```python
# Task 4
# Find the maximum price_per_person for each day and gender combination.
maximum_price = df.groupby(['day', 'sex']).mean(numeric_only=True)
display(maximum_price['price_per_person'].to_frame())
display(f'the maximum_price is -> {maximum_price['price_per_person'].max()}')
```

```python
# Task 5
# Find the minimum price_per_person for each day and size combination.
minimum_price = df.groupby(['day', 'size']).mean(numeric_only=True)
display(minimum_price['price_per_person'].to_frame())
display(f'the minimum_price is -> {minimum_price['price_per_person'].min()}')
```

```python
# Task 6
# Calculate the sum of total_bill grouped by size and smoker status.
# Then use .loc to extract the sum for size 2 smokers only.
Calculate = df.groupby(['size', 'smoker']).mean(numeric_only=True)
print(f'\n The sum of total_bill is -> {Calculate['total_bill'].sum()}\n')
display(Calculate['total_bill'].to_frame())
display(Calculate['total_bill'].loc[[(2,'Yes')]].to_frame())
```

```python
# Task 7
# Display the maximum price_per_person grouped by day and gender. 
# Then use .loc to show the results for Male only across all days.
maximum_price = df.groupby(['day', 'sex']).mean(numeric_only=True)
display(maximum_price['price_per_person'].to_frame())
print(f'\n the maximum price_per_person -> {maximum_price['price_per_person'].max()}\n')
display(maximum_price['price_per_person'].loc[:, 'Male'].to_frame())
```
