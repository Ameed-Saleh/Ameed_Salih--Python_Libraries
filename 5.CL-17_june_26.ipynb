```python
# get last 4 digit from CC Number
tips_df = pd.read_csv('tips.csv')

# apply won't help since it works on 1 column only
#  df['Generous'] = tip/total_bill > 10% ? "Generous" else "Not Generous"

# solution -- np.vectorize() --> get few columns and run a function 
# np.vectorize() = apply with few columns
def is_generous(total_bill, tip):
    if (tip / total_bill) * 100 > 10:
        return 'Generous'
    else:       
        return 'Not Generous'

display(tips_df.head())
tips_df['category'] = np.vectorize(is_generous)(df['total_bill'], tips_df['tip'])
tips_df.tail()
# tips_df[ tips_df['category'] == 'Generous' ]

# create a new column 'table_type'
# if size > 4 then Extra Large else size > 2 then large else if 2 => Regular, else 1 -- Single
# if smoker - Smokers, else Healthy
#   Healthy Regular 
#   Smokers large


def Dimensions(size, smoker):
    match smoker:
        case 'No':
            health_status = 'Healthy'
        case 'Yes':
            health_status = 'smokers'
        case _:
            health_status = 'Unknown'

    match size:
        case 5 | 6: 
            size_status = 'Extra Large'
        case 4 | 3: 
            size_status = 'Large'
        case 2: 
            size_status = 'Regular'
        case _: 
            size_status = 'Single'
            
    return f"{health_status} - {size_status}"
  
display(tips_df.head(5))
tips_df['table_type'] = np.vectorize(Dimensions)(df['size'], tips_df['smoker'])
tips_df.tail(5)
```
