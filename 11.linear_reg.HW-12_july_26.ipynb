```python
import pandas as pd
import numpy as np
```

```python
y      = np.array([10, 12, 20, 18])
y_pred = np.array([11, 13, 18, 18])

sse = np.sum( (y - y_pred)**2 ) 
sst = np.sum( (y - np.mean(y))**2 )
R2 = 1 - (sse / sst)
print(f'sse = {sse}\n')
print(f'sst = {sst}\n')
print(f'R² = {R2}\n') # R2 = 0.9117647058823529
print('a good model always has SSE much smaller than SST ‼️.\n')
print('91.18% of the variance explained. Higher is better..')
```

```python
from sklearn.metrics import r2_score

y      = np.array([10, 12, 20, 18])
y_pred = np.array([11, 13, 18, 18])

R2 = r2_score(y, y_pred)
print(f'R² = {R2}') # R =0.9117647058823529 
print(f'Yes the two numbers are identical -> R² =0.9117647058823529 => 91.18% ')
```

```python
X = np.array([
    [2, 1, 3], [4, 2, 1], [6, 3, 5], [8, 1, 4],
    [3, 5, 2], [5, 4, 6], [7, 2, 3], [9, 5, 1],
    [1, 3, 4], [10, 4, 5]
])

y = np.array([18, 20, 36, 40, 15, 32, 33, 30, 17, 43])

# Use LinearRegression to find and print the intercept and coefficients
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
model = LinearRegression()
model.fit(X, y)

print(f'model.intercept_ = {model.intercept_ }\n')
print(f'model.coef_ = {model.coef_}\n')

# Find the MSE
y_pred = model.predict(X)
MSE = mean_squared_error(y, y_pred)
print(f'MSE = {MSE}\n')

# Find the R² score using r2_score
r2_manual = r2_score(y, y_pred)
print(f'R² = {r2_manual}\n')

# Explain whether the R² value you found is a good score or not
print(f'the R² is excellent because the model explains 98.95% of the variance, which means we have eliminated 98.95% of the unexplained error.')
```
