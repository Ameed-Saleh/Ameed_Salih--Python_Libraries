```python
import io
import numpy as np 
import pandas as pd
from IPython.display import display
```

```python
# Exercise:
# Use the dataset below — a small orders table for a restaurant chain. 
# The data is provided as a CSV string so you don't need any files.
```

```python
# Dataset — paste this into your notebook
csv_data = """order_id,customer,city,total_bill,tip,table_size
1,Alice,Tel Aviv,45.50,8.00,4
2,Bob,Haifa,22.30,3.50,2
3,Carol,Tel Aviv,67.80,12.00,6
4,David,Jerusalem,15.20,2.00,2
5,Eve,Haifa,89.90,15.00,8
6,Frank,Tel Aviv,33.40,5.50,3
7,Grace,Jerusalem,51.00,9.00,4
8,Henry,Haifa,28.60,4.00,2
9,Iris,Tel Aviv,74.20,13.50,5
10,Jake,Jerusalem,19.80,2.50,2"""

# Task 1
# Load csv_data into a DataFrame using io.StringIO and pd.read_csv.
# Print .info() and .describe().transpose()
print(f'---- Task 1 ----')  
df = pd.read_csv(io.StringIO(csv_data))
display(df)    
display(df.info())
display(df.describe().transpose())

```

```python
# Task 2
# Add a column price_per_person that holds the bill divided by the table size, rounded to 2 decimal places.
# Formula: total_bill / table_size
df = pd.read_csv(io.StringIO(csv_data))
print(f'---- Task 2 ----')    
df['price_per_person']= (df['total_bill'] / df['table_size']).round(2)
display(df)

print()
# Task 3
# Filter the DataFrame to show only orders where price_per_person is greater than 12
print(f'---- Task 4 ----') 
display(df[ df['price_per_person'] > 12])
```

```python
# Task 4
# Use .apply() to create a column table_type: 'Large' if table_size >= 4, otherwise 'Small'
print(f'---- Task 4 ----')  
df = pd.read_csv(io.StringIO(csv_data))
def size(table_size):
    if table_size >= 4:
        return 'Large'
    return 'Small'

df['table_type'] = df['table_size'].apply(size)
display(df)
```

```python
# Task 5
# Set customer as the index. Then use .loc[] to select the rows from Alice to Carol
print(f'---- Task 5 ----')   
df = pd.read_csv(io.StringIO(csv_data))
df = df.set_index('customer')
display(df)
display(df.loc['Alice':'Carol'])
```

```python
# Task 6
# Use pd.concat with axis=0 to create a new DataFrame containing the first 3 rows and the last 3 rows of the original DataFrame.
# Reset the index
print(f'---- Task 6 ----')  
df = pd.read_csv(io.StringIO(csv_data))
display(pd.concat([df.head(3), df.tail(3)], axis=0)) # index 0, 1, 2, 7, 8, 9
print()
display(pd.concat([df.head(3), df.tail(3)], axis=0, ignore_index=True)) # index 0, 1, 2, 3, 4, 5
```

```python
# Task 7
# Use np.vectorize to create a column value_label:
# 'High Value' if total_bill > 50 and tip > 10%, otherwise 'Standard'
print(f'---- Task 7 ----')  
df = pd.read_csv(io.StringIO(csv_data))
def value(total_bill, tip):
    if (total_bill < 50) and (tip / total_bill) * 100 > 10:
        return 'High Value'
    return 'Standard'    

df['value_label'] = np.vectorize(value)(df['total_bill'], df['tip'])
display(df)
```

```python
# Task 8
# Use .iloc[] to select the last 3 rows of the DataFrame, then use .iloc[] again to select rows at positions 2, 4, and 6
print(f'---- Task 8 ----')  
df = pd.read_csv(io.StringIO(csv_data))
display(df.iloc[-3:])

display(df.iloc[[2, 4, 6]])

```
