```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from sklearn.cluster import KMeans
from PIL import Image
```

```python
image = Image.open('dog_image.jpg')
image_np = np.array(image)

original_shape = image_np.shape
# reshape to (num_pixels, 3)
# pixels = image_np.reshape(-1, 3)


# task 
# 1 Kmean - clusters: 4, random_state=42
# 2 reshape -- column of pixels 
# 3 kmean.fit 
# print (kmeans.cluster_centers_) == 4
# print (kmeans.labels_) == 240,000
# #1 -- group 0 -- color cluster_centers_[0]  122,50,70
# create new list with center sizes 
# reshape to original picture shape

kmeans = KMeans(n_clusters=4, random_state=42)
image_kmean = kmeans.fit_predict(image_np.reshape(-1, 3))
image_kmean

print(kmeans.cluster_centers_)

print()
print(kmeans.labels_) # 0 means cluster center 0

color_labels = []
for label in kmeans.labels_:
    color_labels.append(kmeans.cluster_centers_[label])

recolored = np.array(color_labels).reshape(original_shape).astype(np.uint8)

k=4

# plot
plt.figure(figsize=(6, 4))
plt.imshow(recolored)
plt.title(f'Image with K={k} colors')
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
image = Image.open('dog_image.jpg')
image_np = np.array(image)

original_shape = image_np.shape
# reshape to (num_pixels, 3)
# pixels = image_np.reshape(-1, 3)


# task 
# 1 Kmean - clusters: [256, 100, 50, 20, 10, 5, 4, 3, 2], random_state=42
# 2 reshape -- column of pixels 
# 3 kmean.fit 
# print (kmeans.cluster_centers_) == 4
# print (kmeans.labels_) == 240,000
# #1 -- group 0 -- color cluster_centers_[0]  122,50,70
# create new list with center sizes 
# reshape to original picture shape

for k in [256, 100, 50, 20, 10, 5, 4, 3, 2]:
        
    kmeans = KMeans(n_clusters=k, random_state=42)
    image_kmean = kmeans.fit_predict(image_np.reshape(-1, 3))
    image_kmean
    
    # print(kmeans.cluster_centers_)
    
    print()
    print(kmeans.labels_) # 0 means cluster center 0
    
    color_labels = []
    for label in kmeans.labels_:
        color_labels.append(kmeans.cluster_centers_[label])
    
    recolored = np.array(color_labels).reshape(original_shape).astype(np.uint8)

    # plot
    plt.figure(figsize=(6, 4))
    plt.imshow(recolored)
    plt.title(f'Image with K={k} colors')
    plt.axis('off')
    plt.tight_layout()
    plt.show()
```
