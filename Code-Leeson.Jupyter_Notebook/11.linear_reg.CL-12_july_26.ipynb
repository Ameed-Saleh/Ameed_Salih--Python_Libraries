```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score


########################################## generate ranodm feature and result based on formula with some 
# 1. Generate synthetic dataset with 3 features
np.random.seed(42)
X = np.random.rand(100, 3) * 10   # 100 samples, 3 features

# Define true relationship: y = 2.5*x1 - 1.5*x2 + 3.0*x3 + 5 + noise
y = 2.5 * X[:, 0] - 1.5 * X[:, 1] + 3.0 * X[:, 2] + 5 + np.random.randn(100) * 2
########################################################################################################

model = LinearRegression()
model.fit(X, y)
print(model.intercept_)
print(model.coef_)

print("--- Model Results ---")
y_pred = model.predict(X)
mse = mean_squared_error(y, y_pred)
print('mse=', mse)

r2 = r2_score(y, y_pred)
print('r2 = ', r2)

sse = np.sum( (y - y_pred)**2 )
sst = np.sum( (y - np.mean(y))**2 )
r2_manual = 1 - (sse/sst)
print('r2_manual', r2_manual)
```
