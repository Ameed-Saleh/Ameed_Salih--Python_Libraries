```python
import pandas as pd
import numpy  as np
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def linear_function(x):
    y = 0.8 * x -2
    return y
x_values = np.linspace(-15, 35, 20)
y_values = linear_function(x_values)
y_values = 0.8 * x_values - 2 # supported in np

# [1, 2, 3]
# 2 * x - 5 
# [-3, -1, 1, ]

plt.axhline(0, color='black',linewidth=0.8, linestyle='--') # X-axis
plt.axvline(0, color='black',linewidth=0.8, linestyle='--') # X-axis
plt.plot(x_values, y_values, color='green')
plt.title('y = 0.8 * x -2')
plt.xlim(-10, 20)
plt.ylim(-20, 20)
plt.grid(color='gray', linestyle='--',linewidth=0.5)
plt.show()
```

```python
# targil
# x + y = 5
# 2x - y = 1
# line 1 - green
# line 2 - red
# add legend: with both equation
# junction point (2,3) -- # bonus use annotate to show the point

def linear_function(x, m, b):
    y = m * x + b
    return y
x1_values = np.linspace(-15, 35, 20)
y1_values = -1 * x1_values + 5 # supported in np   
y2_values = 2 * x1_values - 1 # supported in np   

plt.axhline(0, color='black',linewidth=0.8, linestyle='--') # X-axis
plt.axvline(0, color='black',linewidth=0.8, linestyle='--') # X-axis

plt.plot(x1_values, y1_values, color='green', label='y = -x + 5')
plt.plot(x1_values, y2_values, color='red', label='y = 2x - 1')
plt.legend()
plt.scatter(2, 3, color='blue', zorder=5)

plt.title('targil', color='blue')

plt.xlim(-10, 20)
plt.ylim(-20, 20)
plt.grid(color='gray', linestyle='--',linewidth=0.5)
plt.annotate("(2, 3)", xy=(1, 6), fontsize=10, color='black')
# (    'Junction Point (2,3)', 
#     xy=(2, 3),          # The point to arrow points to
#     xytext=(6, 8),      # Where the text is placed
#     arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6)
# )

plt.show()
```

```python
x_values = np.linspace(-15, 35, 20)
y_values1 = -1 * x_values + 5
y_values2 = 2 * x_values - 1

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # x-axis
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # x-axis
plt.plot(x_values, y_values1, color='green', label='y = -x + 5')
plt.plot(x_values, y_values2, color='red', label='y = 2x - 1')

from sympy import symbols, Eq, solve
x = symbols('x')
eq1 = -x + 5
eq2 = 2 * x - 1
solution = solve( Eq(eq1, eq2), x )
# or:
# solution = solve( Eq(-x + 5, 2 * x - 1), symbols('x') )
x_junc = float(solution[0])  # =2
y_junc = -1 * x_junc + 5
plt.scatter(x_junc, y_junc, color='blue', zorder=5)
plt.annotate(f"({x_junc}, {y_junc})", (x_junc, y_junc), xytext=(x_junc -1.8, y_junc + 5), fontsize=10, color='blue')
# plt.scatter(2, 3, color='blue', zorder=5)
# plt.annotate("(2, 3)", xy=(1, 6), fontsize=10, color='blue')

plt.title('Junction')
plt.xlim(-10, 20)
plt.ylim(-20, 20)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```

```python
'''
draw the 2 lines (with grid, legend...)
sovle with Eq solve + scatter the point + annotate
x + y = 12
x - y = 4
'''

x_values = np.linspace(-15, 35, 20)
y_values1 = -1 * x_values + 12
y_values2 =  x_values - 4

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # x-axis
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # x-axis
plt.plot(x_values, y_values1, color='green', label='y = -x + 12')
plt.plot(x_values, y_values2, color='red', label='y = x - 4')

from sympy import symbols, Eq, solve

solution = solve( Eq(-x + 12, x - 4), symbols('x') )
# or:
# x = symbols('x')
# eq1 = -x + 12
# eq2 = x - 4
# solution = solve( Eq(eq1, eq2), x )

x_junc = float(solution[0])  # =2
y_junc = -1 * x_junc + 12
plt.scatter(x_junc, y_junc, color='blue', zorder=5)
plt.annotate(f"({x_junc}, {y_junc})", (x_junc, y_junc), xytext=(x_junc -2, y_junc + 6), fontsize=12, color='blue')
# plt.scatter(2, 3, color='blue', zorder=5)
# plt.annotate("(2, 3)", xy=(1, 6), fontsize=10, color='blue')

plt.title('Junction')
plt.xlim(-10, 25)
plt.ylim(-20, 25)
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.legend()
plt.show()
```
