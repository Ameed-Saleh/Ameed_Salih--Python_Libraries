```python
X = np.array([1, 2, 3, 4]) # Features - מאפיינים
Y = np.array([1, 2, 3, 4]) # Labels - מה שמחפשים
'''
  X   Y
  1   1
  2   2
  3   3
  4   4

  7   ?
  12  ?
'''

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # x-axis
plt.axvline(0, color='black', linewidth=0.8, linestyle='--')  # y-axis
# plt.plot([1, 2], [1, 2], color='blue', label='?')

# append a and b as tuple to array
# calculate mse per all 4 points - and add to array
# add it to array
# find the best (lowest) mse and print a + b + draw the line 

for i in range(4):
    plt.scatter(X[i], Y[i], zorder=5, s = 200)

ab = []
mse_arr = []

for a in np.arange(-5, +5, 1):
    for b in np.arange(-5, +5, 1):
        ab.append((a, b))
        mse = ((Y - (a * X + b)) ** 2).mean()
        mse_arr.append(mse)
        # print(f'{mse}')
        
mse_np = np.array(mse_arr)
print(f'\n lowest mse {mse_np.min()}')
print(f'\n index lowest {mse_np.argmin()}')
print(f'\n a = {ab[mse_np.argmin()][0]} , b = {ab[mse_np.argmin()][1]}')
# a = ab[mse_np.argmin()][0]
# b = ab[mse_np.argmin()][1]
a, b = ab[mse_np.argmin()]


plt.plot(np.arange(0, 4+1), a * np.arange(0, 4+1) + b, color='red')
plt.title('Linear Regression BF')
plt.grid(color='gray', linestyle='--', linewidth=0.5)
plt.show()
```
