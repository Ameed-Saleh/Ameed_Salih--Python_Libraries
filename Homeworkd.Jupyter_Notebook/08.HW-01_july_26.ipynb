```python
import pandas as pd
import numpy  as np
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("matplotlib_charts_class_exc.csv")
display(df)
display(df.describe().T)
display(df.info())
```

```python
# Task 1
# Create a line chart representing the trend of GDP per Capita over the years for a country of your choice. 
# Ensure the line chart has a title and labelled axes.

df_Germany = df[df['Country'] == 'Germany']
x = df_Germany['Year']
y = df_Germany['GDP per Capita']
plt.plot(x, y, color='red', marker='o')

plt.title("GDP for Germany")
plt.xlabel('Year')
plt.ylabel('GDP per Capita')
plt.legend(['Germany'])
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
# Task 2
# Using a histogram, visualize the distribution of the GDP per Capita values for a particular year of your choice.

df_year = df[df['Year'] == 2002]['GDP per Capita']
plt.hist(df_year, bins=10, color='skyblue', edgecolor='black')

plt.title("Distribution of GDP per Capita (2002)")
plt.xlabel("GDP per Capita")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
# Task 3
# Plot a bar chart to compare the GDP per Capita for five different countries in the year 2010.


df_Year = df[df['Year'] == 2010].head(5)
Countries = df_Year['Country']
GDP = df_Year['GDP per Capita']

bar_colors = ['skyblue', 'gold', 'lightgreen', 'tomato', 'gray']

plt.bar(Countries, GDP , color=bar_colors, edgecolor='black')

plt.xlabel('Country')
plt.ylabel('GDP per Capita')
plt.title('GDP per Capita Comparison (2010)')

plt.tight_layout() 
plt.show()
```

```python
# Task 4
# Using a scatter plot, explore the relationship, if any, between the year and GDP per Capita for a chosen country.

df_Country = df[df['Country'] == 'USA']
Years = df_Country['Year']
GDP = df_Country['GDP per Capita']

plt.scatter(Years, GDP, color='red')
plt.title('GDP per Capita Over Time (USA)')
plt.xlabel('Years')
plt.ylabel('GDP per Capita')
plt.tight_layout()
plt.show()
```
