```python
# Exercise
# The tables below show how many hours each worker worked each day.
# Not every worker appears every day — this is intentional, so you must use .add(fill_value=0).

# Monday
# Worker	Hours
# Alice	     8
# Bob	     6
# David	     7

# Tuesday
# Worker	Hours
# Bob	     8
# Carol	     5
# David	     6

# Wednesday
# Worker	Hours
# Alice    	 7
# Carol	     8
# David	     5
```

```python
# Task 1
# Create a pd.Series for each day using the tables above as your data.

Monday = pd.Series({ 
    'Alice': 8, 
    'Bob': 5,
    'David': 7
})

Tuesday = pd.Series({
    'Bob': 8, 
    'Carol': 5,
    'David': 6
})

Wednesday = pd.Series({
    'Alice': 7, 
    'Carol': 8,
    'David': 5
})
print(f'--- Monday ---\n{Monday}\n')
print(f'--- Tuesday ---\n{Tuesday}\n')
print(f'--- Wednesday ---\n{Wednesday}\n')
```

```python
# Task 2
# Combine all three Series into one total_hours Series showing each worker's total hours across the week.
#Use .add(fill_value=0) so missing days count as 0.

total_hours = Monday.add(Tuesday, fill_value=0).add(Wednesday, fill_value=0)
print(f'--- total_hours ---\n{total_hours.astype(int)}')
```

```python
# Task 3 — Bonus ★
# Using total_hours, find which employee worked the most hours this week.

total_hours = Monday.add(Tuesday, fill_value=0).add(Wednesday, fill_value=0)
print(f'The employee worked the most hours this week is -> {total_hours.idxmax()} -- {total_hours.max()}')
```

```python
# Task 4 — Bonus ★
# Build a DataFrame from the three shift tables — one row per worker, one column per day. Use 0 for days the worker did not work.

work_Table = {'Worker': ['Alice', 'Bob', 'Karol', 'David'],
        'MON': [8, 6, 0, 7],
        'TUE': [0, 8, 5, 6],
        'WED': [7, 0, 8, 5]}

df = pd.DataFrame(work_Table)

print(df)
```
